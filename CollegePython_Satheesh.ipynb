Student attendance management system

```python
Students = ['Fayas', 'Dhyan', 'Varun', 'Dilshan', 'Aslah', 'Sananjay', 'Arif', 'Martin', 'Nasheeb', 'andriya', 'Akshitha', 'Fariz', 'Akshay',
            'Abhinav', 'Shreyas', 'Dakshim', 'Niranjana', 'Athul', 'Sona', 'Anamika', 'Dhanusree', 'Abhishek', 'Hari']

present = 0
absent = 0

for student in Students:
  attendance = input('Is ' + student + ' present? (Yes/No) : ')

  if attendance == 'yes':
    print(student, '- Present')
    present = present + 1
  else:
    print(student, '- Absent')
    absent = absent + 1


print("\nTotal Present:", present)
print("Total Absent:", absent)
```
