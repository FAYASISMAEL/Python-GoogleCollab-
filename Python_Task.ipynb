**Task 3 — Resume Generator · Ask for name, education, skills, projects, languages → print a clean professional resume.**

```python
print('=' * 60)
print("                 RESUME GENERATOR")
print('=' * 60)

name = input('Enter your full name : ')
education = input('Enter your education : ')
skills = input('Enter your skills (separated by comas) : ')
projects = input('Enter your project (separated by commas) : ')
languages = input('Enter programming languages you know (separated by commas) : ')

skills_list = skills.split(',')
projects_list = projects.split(',')
languages_list = languages.split(',')

print('\n\n')
print('=' * 60)
print(f"                    {name.upper()}")
print('=' * 60)

print("\nEDUCATION")
print("-" * 60)
print(education)

print("\nSKILLS")
print("-" * 60)

for skill in skills_list:
    print(f"• {skill.strip()}")

print("\nPROJECTS")
print("-" * 60)

for project in projects_list:
    print(f"• {project.strip()}")

print("\nPROGRAMMING LANGUAGES")
print("-" * 60)

for language in languages_list:
    print(f"• {language.strip()}")

print("\n" + "=" * 60)
print("                 END OF RESUME")
print("=" * 60)
```

**Task - 04 - Portfolio README Generator · Generate a ready-to-paste GitHub README.md from the user's details. (Learn Markdown here.)italicized text**

```python

print('=' * 60)
print("             PORTFOLIO README GENERATOR")
print('=' * 60)

name = input('Enter your name : ')
title = input('Enter your professional title : ')
about = input('Write a short description about yourself : ')

skills = input('Enter your Skills : ')
projects = input('Enter your projects : ')

github = input('Enter your Github URL : ')
linkedin = input('Enter your LinkedIn URL : ')
email = input('Enter your email : ')

skills_list = skills.split(',')
projects_list = projects.split(',')

skills_markdown = ''

for skill in skills_list:
  skills_markdown += f'- {skills.strip()}\n'

projects_markdown = ''

for project in projects_list:
  projects_markdown += f'- **{project.strip()}**\n'

readme_content = f""\

{title}
{about}

{skills_markdown}
{projects_markdown}

Email: {email}
Github: {github}
LinkedIn: {linkedin}

# with open('README.md', 'w', encoding='utf-8') as file:
#   file.write(readme_content)

# print("===== README.md FILE CONTENT =====")
# print(readme_content)

# print('\n')
# print('=' *60)
# print("             GENERATED README.md")
# print("=" * 60)

# print("=" * 60)
# print("README.md has been created successfully!")
# print("=" * 60)

filename = "README.md"

with open(filename, "w", encoding="utf-8") as file:
    file.write(readme_content)

print("README.md has been created successfully!")

# Verify the file
with open(filename, "r", encoding="utf-8") as file:
    saved_content = file.read()

print("\n===== SAVED README.md =====")
print(saved_content)

# File download
from google.colab import files
files.download(filename)
```
